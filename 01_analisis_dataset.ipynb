## 5.3 Resampling Time Series

untuk membentuk deret waktu dengan interval yang seragam



```python
# Menjadikan updated_at sebagai index
df_device = df_device.set_index("updated_at")

# Resampling harian
daily_aqi = (
    df_device["aqi"]
    .resample("D")
    .mean()
)

daily_aqi.head()
```

```python
print(daily_aqi.shape)
```

```python
plt.figure(figsize=(15,5))

plt.plot(
    daily_aqi.index,
    daily_aqi.values,
    linewidth=2
)

plt.title("Rata-rata AQI Harian")
plt.xlabel("Tanggal")
plt.ylabel("AQI")

plt.grid(True)

plt.show()
```

### Interpretasi

Visualisasi deret waktu hasil resampling harian menunjukkan bahwa nilai Air Quality Index (AQI) mengalami fluktuasi sepanjang periode pengamatan, mulai dari Juni 2025 hingga Juni 2026. Nilai AQI tidak berada pada tingkat yang konstan, melainkan mengalami perubahan dari waktu ke waktu dengan tingkat variasi yang cukup tinggi.

Secara umum, tidak terlihat pola kenaikan atau penurunan yang bersifat linier dalam jangka panjang. Namun demikian, pada beberapa periode, khususnya menjelang pertengahan tahun 2026, terlihat adanya peningkatan nilai AQI dibandingkan periode sebelumnya. Kondisi ini mengindikasikan adanya kecenderungan perubahan pola kualitas udara yang perlu dianalisis lebih lanjut melalui identifikasi tren dan pola musiman.

Selain itu, grafik juga memperlihatkan beberapa lonjakan nilai AQI yang relatif tinggi pada waktu tertentu. ## 5.6 Penanganan Missing Date

untuk membentuk deret waktu yang kontinu dengan mengisi tanggal yang tidak memiliki hasil pengukuran

menangani waktu yang kosong ada beberapa solusi
1. menghapus tanggal kosong, namun jarak waktu nanti tidak sama, padahal sarima menganggap data memiliki interval yang sama
2. Forward Fill adalah menyamakan data yang hilang ke data sebelumnya, kurang cocok karena AQI berubah setiap hari

11 Januari = NaN

## 5.8 Analisis Komponen Deret Waktu

untuk memisahkan deret waktu Air Quality Index (AQI) ke dalam komponen trend, musiman (*seasonality*), dan residual menggunakan metode Seasonal Decomposition

Mengapa period=7?

untuk data harian, nilai period harus mencerminkan 1 siklus yang ingin diuji, nilai 7 digunakan untuk melihat apakah pola yang berulan setiap minggu (7 hari).

setelah melihat hasil dekomposisi, dapat mengevaluasi apakah pola mingguan terlihat. jika tidak bisa mencoba period 30 (30 hari/bulanan) 

```python
from statsmodels.tsa.seasonal import seasonal_decompose
import matplotlib.pyplot as plt

decomposition = seasonal_decompose(
    daily_df["aqi"],
    model="additive",
    period=7
)

fig = decomposition.plot()
fig.set_size_inches(15,10)

plt.show()
```

### Interpretasi

Hasil *Seasonal Decomposition* menunjukkan bahwa deret waktu Air Quality Index (AQI) dapat dipisahkan menjadi empat komponen utama, yaitu data aktual (*observed*), komponen tren (*trend*), komponen musiman (*seasonal*), dan komponen residual (*residual*).

Komponen tren memperlihatkan adanya perubahan nilai AQI sepanjang periode pengamatan. Pada awal pengamatan hingga sekitar Oktober 2025, nilai AQI cenderung berada pada tingkat yang lebih tinggi, kemudian mengalami penurunan pada akhir tahun 2025. Selanjutnya, mulai awal tahun 2026 hingga akhir periode pengamatan terlihat adanya kecenderungan peningkatan kembali. Komponen musiman menunjukkan adanya pola yang berulang secara periodik dengan amplitudo yang relatif stabil. Hal ini mengindikasikan bahwa terdapat pengaruh musiman terhadap perubahan nilai AQI, meskipun besarnya variasi musiman relatif lebih kecil dibandingkan fluktuasi data aktual.

Sementara itu, komponen residual memperlihatkan penyebaran nilai yang berfluktuasi di sekitar nol tanpa membentuk pola tertentu. Kondisi tersebut menunjukkan bahwa sebagian besar pola sistematis pada data telah berhasil dipisahkan ke dalam komponen tren dan musiman, sedangkan residual merepresentasikan variasi acak yang tidak dapat dijelaskan oleh kedua komponen tersebut.

## 5.9 Uji Stasioneritas (Augmented Dickey-Fuller Test)

untuk menguji apakah deret waktu Air Quality Index (AQI) telah memenuhi sifat stasioner

Hipotesis ADF

Keputusan:

p-value < 0,05 → Tolak H0 → Data stasioner

p-value ≥ 0,05 → Gagal menolak H0 → Data tidak stasioner

```python
from statsmodels.tsa.stattools import adfuller

result = adfuller(daily_df["aqi"])

print("ADF Statistic :", result[0])
print("p-value       :", result[1])
print("Lag Used      :", result[2])
print("Observations  :", result[3])

print("\nCritical Values")

for key, value in result[4].items():
    print(f"{key} : {value}")
```

ADF Statistic (-7,7888) lebih kecil (lebih negatif) daripada seluruh nilai kritis (1%, 5%, dan 10%).
p-value jauh lebih kecil dari 0,05.

```python
alpha = 0.05

print(f"p-value = {result[1]:.6f}")

if result[1] < alpha:
    print("\nKeputusan : Tolak H0")
    print("Kesimpulan : Data stasioner")
else:
    print("\nKeputusan : Gagal menolak H0")
    print("Kesimpulan : Data belum stasioner")
```

H₀ ditolak, sehingga deret waktu AQI harian dapat dinyatakan stasioner berdasarkan uji Augmented Dickey-Fuller.

### Interpretasi

Pengujian stasioneritas menggunakan metode Augmented Dickey-Fuller (ADF) menghasilkan nilai **ADF Statistic sebesar -7,7888** dengan **p-value sebesar 8,05 × 10⁻¹²**. Nilai tersebut lebih kecil dari taraf signifikansi 0,05, sehingga hipotesis nol (H₀) yang menyatakan bahwa deret waktu mengandung unit root dapat ditolak.

Selain itu, nilai ADF Statistic juga lebih kecil dibandingkan seluruh nilai kritis pada tingkat signifikansi 1%, 5%, dan 10%. Hasil ini menunjukkan bahwa deret waktu Air Quality Index (AQI) harian telah memenuhi sifat stasioner.

Dengan demikian, berdasarkan hasil uji ADF, data tidak memerlukan proses differencing tambahan (d = 0). Oleh karena itu, tahap selanjutnya adalah mengidentifikasi struktur autokorelasi menggunakan grafik Autocorrelation Function (ACF) dan Partial Autocorrelation Function (PACF) sebagai dasar penentuan parameter awal model SARIMA.

## 5.10 Analisis Autocorrelation Function (ACF)

 untuk mengidentifikasi hubungan antara nilai Air Quality Index (AQI) pada suatu waktu dengan nilai AQI pada beberapa periode sebelumnya (lag)

```python
from statsmodels.graphics.tsaplots import plot_acf
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

plot_acf(
    daily_df["aqi"],
    lags=40,
    ax=plt.gca()
)

plt.title("Autocorrelation Function (ACF)")

plt.show()
```

## 5.11 Analisis Partial Autocorrelation Function (PACF)

untuk mengidentifikasi hubungan langsung antara nilai Air Quality Index (AQI) dengan nilai pada beberapa lag sebelumnya setelah menghilangkan pengaruh lag di antaranya

```python
from statsmodels.graphics.tsaplots import plot_pacf

plt.figure(figsize=(12,5))

plot_pacf(
    daily_df["aqi"],
    lags=40,
    method="ywm",
    ax=plt.gca()
)

plt.title("Partial Autocorrelation Function (PACF)")

plt.show()
```

Lag adalah selang waktu ke belakang yang digunakan untuk membandingkan nilai suatu data dengan nilai pada waktu sebelumnya.

### Interpretasi 

Grafik ACF menunjukkan nilai autokorelasi yang tinggi pada lag awal dan kemudian menurun secara bertahap (tailing off), yang mengindikasikan bahwa hubungan antarobservasi melemah seiring bertambahnya lag. Sementara itu, grafik PACF memperlihatkan spike yang dominan pada lag pertama, sedangkan lag berikutnya sebagian besar berada dalam batas kepercayaan. Hasil ini menunjukkan bahwa komponen autoregressive diperkirakan berorde rendah, sementara penentuan parameter akhir model SARIMA tetap dilakukan melalui evaluasi beberapa kombinasi parameter menggunakan kriteria AIC/BIC dan performa prediksi.

## 6.1 Periksa Jumlah Data Setiap Perangkat 

```python
# Memeriksa jumlah data dan rentang waktu setiap perangkat

ringkasan_perangkat = (
    df.groupby("device")
    .agg(
        jumlah_pengukuran=("aqi", "count"),
        tanggal_awal=("updated_at", "min"),
        tanggal_akhir=("updated_at", "max")
    )
    .sort_values("jumlah_pengukuran", ascending=False)
)

display(ringkasan_perangkat.head(15))
```

## 6.2 Memeriksa Kelengkapan data harian

```python
# Memeriksa kelengkapan data harian pada 15 perangkat teratas

perangkat_teratas = ringkasan_perangkat.head(15).index

data_valid = df[
    df["device"].isin(perangkat_teratas)
    & df["aqi"].notna()
].copy()

data_valid["tanggal"] = data_valid["updated_at"].dt.normalize()

kelengkapan_harian = (
    data_valid.groupby("device")
    .agg(
        jumlah_pengukuran=("aqi", "count"),
        tanggal_awal=("tanggal", "min"),
        tanggal_akhir=("tanggal", "max"),
        hari_teramati=("tanggal", "nunique")
    )
)

kelengkapan_harian["total_hari"] = (
    kelengkapan_harian["tanggal_akhir"]
    - kelengkapan_harian["tanggal_awal"]
).dt.days + 1

kelengkapan_harian["hari_hilang"] = (
    kelengkapan_harian["total_hari"]
    - kelengkapan_harian["hari_teramati"]
)

kelengkapan_harian["cakupan_persen"] = (
    kelengkapan_harian["hari_teramati"]
    / kelengkapan_harian["total_hari"] * 100
).round(2)

display(
    kelengkapan_harian.sort_values(
        "cakupan_persen", ascending=False
    )
)
```

## 6.3 Membuat Data AQI Harian


```python
# Langkah 4: Membuat seri AQI harian

nama_perangkat = "Jl. Lapangan Rajawali"

data_perangkat = df[
    (df["device"] == nama_perangkat)
    & (df["aqi"].notna())
].copy()

# Mengubah waktu menjadi tanggal harian
data_perangkat["tanggal"] = (
    data_perangkat["updated_at"].dt.normalize()
)

# Menghitung rata-rata AQI setiap hari
aqi_harian = (
    data_perangkat.groupby("tanggal")["aqi"]
    .mean()
    .sort_index()
    .to_frame(name="aqi")
)

print("Nama perangkat:", nama_perangkat)
print("Jumlah hari:", len(aqi_harian))
print("\nLima data AQI harian pertama:")
display(aqi_harian.head())

print("\nLima data AQI harian terakhir:")
display(aqi_harian.tail())
```

Misalnya, dalam satu hari terdapat tiga pengukuran AQI sebesar 60, 70, dan 80. Rata-rata AQI hari itu menjadi:
AQI harian= 60+70+80/3= 70

```python
# Langkah 5: Memeriksa tanggal yang hilang

rentang_tanggal = pd.date_range(
    start=aqi_harian.index.min(),
    end=aqi_harian.index.max(),
    freq="D"
)

tanggal_hilang = rentang_tanggal.difference(aqi_harian.index)

print("Jumlah hari dalam rentang:", len(rentang_tanggal))
print("Jumlah hari teramati:", len(aqi_harian))
print("Jumlah tanggal hilang:", len(tanggal_hilang))

print("\nDaftar tanggal yang hilang:")
print(tanggal_hilang)
```

## 6.5 Membuat seri harian lengkap tanpa mengisi nilai kosong
Tujuan langkah ini adalah menampilkan seluruh tanggal dalam periode penelitian, termasuk tanggal yang tidak memiliki pengukuran. Nilai AQI yang hilang akan tetap NaN untuk sementara.

```python
# Langkah 6: Membuat indeks tanggal harian lengkap

aqi_harian_lengkap = aqi_harian.asfreq("D")

print("Jumlah hari setelah melengkapi indeks:",
      len(aqi_harian_lengkap))

print("Jumlah AQI yang kosong:",
      aqi_harian_lengkap["aqi"].isna().sum())

print("\nTanggal dan nilai AQI yang kosong:")
display(
    aqi_harian_lengkap[
        aqi_harian_lengkap["aqi"].isna()
    ]
)
```

## 6.6 Memeriksa rentang tanggal yang hilang

```python
# Langkah 7: Mengelompokkan tanggal yang hilang secara berurutan

tanggal_kosong = aqi_harian_lengkap[
    aqi_harian_lengkap["aqi"].isna()
].index

kelompok_hilang = (
    pd.Series(tanggal_kosong)
    .diff()
    .dt.days
    .ne(1)
    .cumsum()
)

rentang_hilang = (
    pd.DataFrame({
        "tanggal": tanggal_kosong,
        "kelompok": kelompok_hilang.values
    })
    .groupby("kelompok")
    .agg(
        tanggal_mulai=("tanggal", "min"),
        tanggal_akhir=("tanggal", "max"),
        jumlah_hari=("tanggal", "count")
    )
    .reset_index(drop=True)
)

display(rentang_hilang)

print(
    "Rentang kehilangan terpanjang:",
    rentang_hilang["jumlah_hari"].max(),
    "hari"
)
```

## 6.7 Memeriksa karakteristik AQI di sekitar tanggal yang hilang

```python
# Langkah 8: Memeriksa AQI di sekitar tanggal yang hilang

for _, baris in rentang_hilang.iterrows():
    mulai = baris["tanggal_mulai"]
    akhir = baris["tanggal_akhir"]

    awal_pemeriksaan = mulai - pd.Timedelta(days=3)
    akhir_pemeriksaan = akhir + pd.Timedelta(days=3)

    print(f"\nRentang hilang: {mulai.date()} sampai {akhir.date()}")

    display(
        aqi_harian_lengkap.loc[
            awal_pemeriksaan:akhir_pemeriksaan
        ]
    )
```

## 6.8 Mencari periode data yang berurutan tanpa celah

```python
# Memeriksa rentang data AQI yang lengkap dan berurutan

data_tersedia = aqi_harian_lengkap["aqi"].notna()

kelompok_data = (
    data_tersedia.ne(data_tersedia.shift())
    .cumsum()
)

rentang_lengkap = (
    aqi_harian_lengkap[data_tersedia]
    .groupby(kelompok_data[data_tersedia])
    .agg(
        tanggal_mulai=("aqi", lambda x: x.index.min()),
        tanggal_akhir=("aqi", lambda x: x.index.max()),
        jumlah_hari=("aqi", "count")
    )
    .sort_values("jumlah_hari", ascending=False)
)

display(rentang_lengkap)
```

## 6.13 Melatih kandidat pertama

```python
# Langkah 17: Membandingkan kandidat ARIMA awal

from statsmodels.tsa.statespace.sarimax import SARIMAX
import pandas as pd

kandidat = {
    "ARIMA(1,0,0)": (1, 0, 0),
    "ARIMA(0,0,1)": (0, 0, 1),
    "ARIMA(1,0,1)": (1, 0, 1)
}

hasil_model = []

for nama, order in kandidat.items():
    model = SARIMAX(
        train["aqi"],
        order=order,
        seasonal_order=(0, 0, 0, 0),
        trend="c",
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    hasil = model.fit(disp=False)

    hasil_model.append({
        "Model": nama,
        "AIC": hasil.aic,
        "BIC": hasil.bic
    })

tabel_model = pd.DataFrame(hasil_model)
tabel_model = tabel_model.sort_values("AIC")

display(tabel_model)
```

ARIMA(1,0,1) merupakan kandidat terbaik berdasarkan AIC dan BIC pada ketiga model yang diuji. Nilai AIC dan BIC yang lebih rendah menunjukkan kompromi yang lebih baik antara kecocokan model dan kompleksitasnya, dibandingkan kandidat lain dalam tabel tersebut.

## 6.14 Menguji prediksi pada data testing

```python
# Langkah 19: Prediksi ARIMA(1,0,1) pada data testing

from statsmodels.tsa.statespace.sarimax import SARIMAX
import pandas as pd

model_arima = SARIMAX(
    train["aqi"],
    order=(1, 0, 1),
    seasonal_order=(0, 0, 0, 0),
    trend="c",
    enforce_stationarity=False,
    enforce_invertibility=False
)

hasil_arima = model_arima.fit(disp=False)

prediksi_arima = hasil_arima.forecast(
    steps=len(test)
)

hasil_prediksi = pd.DataFrame({
    "AQI_Aktual": test["aqi"],
    "AQI_Prediksi": prediksi_arima
})

display(hasil_prediksi)
```

## 6.15 Menghitung MAE,RMSE, MAPE

```python
# Langkah 20: Evaluasi model ARIMA(1,0,1)

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

aktual = hasil_prediksi["AQI_Aktual"]
prediksi = hasil_prediksi["AQI_Prediksi"]

mae = mean_absolute_error(aktual, prediksi)

rmse = np.sqrt(
    mean_squared_error(aktual, prediksi)
)

mape = np.mean(
    np.abs((aktual - prediksi) / aktual)
) * 100

print("Evaluasi ARIMA(1,0,1)")
print("---------------------")
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"MAPE : {mape:.2f}%")
```

MAE = 12,5757 berarti rata-rata selisih absolut prediksi terhadap AQI aktual sekitar 12,58 poin AQI.

RMSE = 13,9512 menunjukkan besarnya kesalahan prediksi dengan penalti lebih besar terhadap kesalahan yang besar. Nilainya yang lebih tinggi daripada MAE mengindikasikan adanya sebagian kesalahan yang relatif besar.

MAPE = 205,41% terlihat sangat tinggi. MAPE membagi kesalahan dengan nilai aktual, sehingga ketika AQI aktual rendah misalnya 1,64 atau 3,67 kesalahan yang relatif kecil sekalipun bisa menghasilkan persentase sangat besar.
